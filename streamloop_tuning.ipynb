# StreamLoop — Ajuste del modelo de cancelación (churn)

Objetivo: partir de un clasificador con hiperparámetros por defecto y ajustarlo de forma sistemática
(`RandomizedSearchCV` amplio → `GridSearchCV` focalizado), optimizando una métrica alineada con el negocio
y revisando la estabilidad entre folds antes de elegir el modelo final.

**Reglas que sigue este notebook**
- El split train/test se hace antes de cualquier modelado.
- Todo el preprocesamiento vive dentro de un `Pipeline`, así que la validación cruzada nunca filtra información entre folds.
- Las búsquedas solo ven `X_train`. El test set se usa **exactamente dos veces**: línea base y modelo final.

```python
import json
import numpy as np
import pandas as pd
from scipy.stats import randint

from sklearn.model_selection import train_test_split, StratifiedKFold, RandomizedSearchCV, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (make_scorer, fbeta_score, recall_score, precision_score, f1_score,
                             roc_auc_score, accuracy_score, confusion_matrix)

RANDOM_STATE = 42
pd.set_option("display.width", 200)
pd.set_option("display.max_columns", 30)
```

## 1. Carga y limpieza mínima

```python
URL = "https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv"
df = pd.read_csv(URL)
print(df.shape)
df.head()
```

```python
# customerID es un identificador: no aporta señal y el modelo podría memorizarlo.
df = df.drop(columns="customerID")

# TotalCharges viene como texto; los clientes con tenure=0 tienen un espacio en blanco en lugar de número.
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"].str.strip().replace("", np.nan), errors="coerce")
print("TotalCharges vacíos:", df["TotalCharges"].isna().sum())

# Objetivo binario: 1 = canceló
y = (df.pop("Churn") == "Yes").astype(int)
X = df
print("Tasa de churn:", round(y.mean(), 3))
```

Los 11 valores vacíos de `TotalCharges` **no** se imputan aquí: la imputación se hace dentro del pipeline
(`SimpleImputer`), para que la mediana se calcule solo con los datos de entrenamiento de cada fold.
Las columnas categóricas (texto) se codifican con `OneHotEncoder`, también dentro del pipeline.

## 2. División train/test (antes de cualquier modelado)

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=RANDOM_STATE
)
print(X_train.shape, X_test.shape)
print("Churn train:", round(y_train.mean(), 3), "| Churn test:", round(y_test.mean(), 3))
```

## 3. Pipeline (preprocesamiento + clasificador)

```python
num_cols = ["tenure", "MonthlyCharges", "TotalCharges"]
cat_cols = [c for c in X.columns if c not in num_cols]   # incluye SeniorCitizen (0/1)

preprocess = ColumnTransformer([
    ("num", Pipeline([
        ("impute", SimpleImputer(strategy="median")),
        ("scale", StandardScaler()),
    ]), num_cols),
    ("cat", OneHotEncoder(handle_unknown="ignore"), cat_cols),
])

def make_pipeline():
    return Pipeline([
        ("pre", preprocess),
        ("clf", RandomForestClassifier(random_state=RANDOM_STATE)),
    ])

make_pipeline()
```

## 4. Métrica de negocio

Para StreamLoop un **falso negativo** (cliente que cancela sin ser detectado) cuesta mucho más que un
**falso positivo** (oferta de retención innecesaria). Eso descarta `accuracy` (con ~27 % de churn, un modelo
que dice "nadie cancela" ya tiene ~73 % de accuracy).

Tampoco optimizamos `recall` a secas: el buscador encontraría trivialmente un modelo que marca a todos como
churn (recall = 1) y la campaña de retención se volvería inútil y cara.

Usamos **F2** (`fbeta_score` con `beta=2`): combina precisión y recall pero da al recall **4 veces más peso**
(β² = 4). Así la búsqueda premia detectar a quienes cancelan, sin dejar de castigar la avalancha de falsos
positivos. Recall, precisión, F1, ROC-AUC y accuracy se reportan como métricas complementarias.

```python
f2_scorer = make_scorer(fbeta_score, beta=2)

def evaluate(model, X_, y_):
    pred = model.predict(X_)
    proba = model.predict_proba(X_)[:, 1]
    tn, fp, fn, tp = confusion_matrix(y_, pred).ravel()
    return {
        "F2": fbeta_score(y_, pred, beta=2),
        "Recall": recall_score(y_, pred),
        "Precision": precision_score(y_, pred),
        "F1": f1_score(y_, pred),
        "ROC-AUC": roc_auc_score(y_, proba),
        "Accuracy": accuracy_score(y_, pred),
        "TP": tp, "FN": fn, "FP": fp, "TN": tn,
    }
```

## 5. Línea base — hiperparámetros por defecto (uso #1 del test set)

```python
baseline = make_pipeline()
baseline.fit(X_train, y_train)
baseline_metrics = evaluate(baseline, X_test, y_test)
pd.Series(baseline_metrics, name="Línea base").round(4)
```

El Random Forest por defecto tiene una accuracy decente pero **deja escapar a más de la mitad de los clientes
que cancelan** (recall bajo). Para el problema de negocio eso es justamente lo que hay que corregir.

## 6. Búsqueda amplia — `RandomizedSearchCV`

Espacio definido con hiperparámetros reales de `RandomForestClassifier`:

| Hiperparámetro | Rango | Por qué |
|---|---|---|
| `n_estimators` | 100–600 | más árboles = menos varianza, con rendimientos decrecientes |
| `max_depth` | None, 4–20 | controla sobreajuste |
| `min_samples_split` | 2–30 | regulariza divisiones |
| `min_samples_leaf` | 1–30 | hojas más grandes = probabilidades más suaves |
| `max_features` | sqrt, log2, 0.3, 0.5, 0.7 | decorrelación entre árboles |
| `class_weight` | None, balanced, balanced_subsample | compensa el desbalance (27 % churn) — clave para el recall |

Validación cruzada estratificada de 5 folds, solo sobre `X_train`.

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

random_space = {
    "clf__n_estimators": randint(100, 601),
    "clf__max_depth": [None, 4, 6, 8, 10, 12, 16, 20],
    "clf__min_samples_split": randint(2, 31),
    "clf__min_samples_leaf": randint(1, 31),
    "clf__max_features": ["sqrt", "log2", 0.3, 0.5, 0.7],
    "clf__class_weight": [None, "balanced", "balanced_subsample"],
}

random_search = RandomizedSearchCV(
    make_pipeline(), random_space, n_iter=60, scoring=f2_scorer,
    cv=cv, n_jobs=-1, random_state=RANDOM_STATE,
)
random_search.fit(X_train, y_train)   # solo datos de entrenamiento
print("Mejor F2 (CV):", round(random_search.best_score_, 4))
random_search.best_params_
```

```python
param_cols = [c for c in random_search.cv_results_ if c.startswith("param_")]
rs_results = (pd.DataFrame(random_search.cv_results_)
              .sort_values("rank_test_score")
              [param_cols + ["mean_test_score", "std_test_score"]])
rs_results.columns = [c.replace("param_clf__", "") for c in rs_results.columns]
rs_results.head(15).round(4)
```

```python
# ¿Qué valores aparecen en el top 10 vs. el bottom 10? Esto guía el acotamiento.
top10, bottom10 = rs_results.head(10), rs_results.tail(10)
for col in ["class_weight", "max_depth", "max_features"]:
    print(f"{col:13s} top10: {top10[col].fillna('None').astype(str).value_counts().to_dict()}")
    print(f"{'':13s} bot10: {bottom10[col].fillna('None').astype(str).value_counts().to_dict()}")
print()
print(top10[["min_samples_leaf", "min_samples_split", "n_estimators"]].describe().loc[["min", "50%", "max"]])
```

**Lectura de la búsqueda aleatoria → cómo acotamos la grilla**
- `class_weight` es lo que más pesa: **todo** el top 10 usa pesos balanceados (8 `balanced`, 2 `balanced_subsample`) y los **10 peores son todos `class_weight=None`** — sin pesos el modelo apenas detecta churn. Fijamos `balanced`. (Como el fondo lo explica entero el peso de clases, para los demás parámetros miramos sobre todo el top.)
- Los mejores árboles son **poco profundos**: 7 de los 10 mejores tienen `max_depth` 4–6, y los tres primeros usan 6. Los profundos (12, 20) solo entran al top cuando hojas muy grandes (`min_samples_leaf` ≥ 20) los regularizan → la grilla explora 5–8.
- `max_features`: `sqrt` (5) y `0.3` (3) dominan el top 10; `log2` y `0.7` no aparecen. Probamos `sqrt` y `0.3`.
- `min_samples_split` alto (mediana 22 en el top 10) y `min_samples_leaf` muy variable → grilla `split ∈ {10, 20, 30}`, `leaf ∈ {1, 3, 6, 10}`.
- `n_estimators` casi no mueve la métrica (top 10 entre 162 y 540) → lo fijamos en 400 para gastar el presupuesto en lo que sí importa.

Resultado: 4 × 2 × 4 × 3 = **96 candidatos** × 5 folds, todos en la región prometedora.

## 7. Búsqueda focalizada — `GridSearchCV`

```python
grid_space = {
    "clf__class_weight": ["balanced"],
    "clf__n_estimators": [400],
    "clf__max_depth": [5, 6, 7, 8],
    "clf__max_features": ["sqrt", 0.3],
    "clf__min_samples_leaf": [1, 3, 6, 10],
    "clf__min_samples_split": [10, 20, 30],
}

grid_search = GridSearchCV(
    make_pipeline(), grid_space, scoring=f2_scorer, cv=cv, n_jobs=-1,
    # refit=True (por defecto): reentrena automáticamente el mejor candidato sobre todo X_train
)
grid_search.fit(X_train, y_train)   # solo datos de entrenamiento
print("Candidatos:", len(grid_search.cv_results_["params"]))
print("Mejor F2 (CV):", round(grid_search.best_score_, 4))
grid_search.best_params_
```

## 8. Revisión de `cv_results_`: promedio **y** variación entre folds

```python
res = pd.DataFrame(grid_search.cv_results_)
fold_cols = [f"split{i}_test_score" for i in range(cv.get_n_splits())]
res["min_fold"] = res[fold_cols].min(axis=1)
res["max_fold"] = res[fold_cols].max(axis=1)
res["rango"] = res["max_fold"] - res["min_fold"]

show = ["rank_test_score", "param_clf__max_depth", "param_clf__max_features",
        "param_clf__min_samples_leaf", "param_clf__min_samples_split",
        "mean_test_score", "std_test_score", "min_fold", "max_fold", "rango"] + fold_cols
top = res.sort_values("rank_test_score")[show].head(10).copy()
top.columns = [c.replace("param_clf__", "") for c in top.columns]
top.round(4)
```

```python
# Regla de "una desviación estándar": los candidatos cuyo promedio está dentro de 1 std del mejor
# son prácticamente indistinguibles con 5 folds.
best = res.loc[grid_search.best_index_]
threshold = best["mean_test_score"] - best["std_test_score"]
tied = res[res["mean_test_score"] >= threshold]
print(f"Mejor promedio: {best['mean_test_score']:.4f} ± {best['std_test_score']:.4f}")
print(f"Candidatos dentro de 1 std del mejor: {len(tied)} de {len(res)}")
print(f"Std del mejor: {best['std_test_score']:.4f} | std mediana de los empatados: {tied['std_test_score'].median():.4f}")
most_stable = tied.loc[tied["std_test_score"].idxmin()]
print("\nMás estable dentro del empate:")
print(f"  params: {most_stable['params']}")
print(f"  F2 medio: {most_stable['mean_test_score']:.4f} ± {most_stable['std_test_score']:.4f} | peor fold: {most_stable['min_fold']:.4f}")
print(f"Mejor candidato | peor fold: {best['min_fold']:.4f}")
```

**Elección del modelo final**

Con 5 folds, la variación entre folds (std ≈ 0.013–0.017, rango ≈ 0.04 entre el mejor y el peor fold) es **mucho mayor** que la diferencia entre los primeros candidatos (el top 10 está separado por menos de 0.003 de F2). Casi la totalidad de la grilla (93 de 96) cae dentro de 1 std del mejor: estadísticamente son un empate.

Por eso no basta con mirar el promedio. Comparamos el mejor contra el candidato más estable del empate:

| Candidato | F2 medio (CV) | Std | Peor fold |
|---|---|---|---|
| Mejor promedio: depth 6, sqrt, leaf 3, split 30 | **0.7292** | 0.0144 | **0.7057** |
| Más estable: depth 7, sqrt, leaf 10, split 30 | 0.7226 | **0.0128** | 0.7027 |

El candidato "más estable" solo reduce la std en 0.0016, pero pierde 0.0066 de promedio **y tiene un peor fold más bajo**. Es decir, su menor std no se traduce en un piso más seguro. El mejor promedio, en cambio, tiene una std por debajo de la mediana del top 10 y el mejor peor-fold del grupo, y además está fuertemente regularizado (profundidad 6, `min_samples_split=30`).

**Decisión:** nos quedamos con el candidato de mayor promedio, que es precisamente el `best_estimator_` que `refit=True` ya reentrenó sobre todo `X_train`. No se reentrena nada a mano.

## 9. Evaluación final en el test set (uso #2 del test set)

```python
final_model = grid_search.best_estimator_   # ya reentrenado por refit=True; no se vuelve a entrenar
final_metrics = evaluate(final_model, X_test, y_test)

comparison = pd.DataFrame({"Línea base": baseline_metrics, "Ajustado": final_metrics})
comparison["Δ"] = comparison["Ajustado"] - comparison["Línea base"]
comparison.round(4)
```

```python
# Guardar resultados para el reporte
summary = {
    "baseline_test": {k: float(v) for k, v in baseline_metrics.items()},
    "tuned_test": {k: float(v) for k, v in final_metrics.items()},
    "random_best_cv_f2": float(random_search.best_score_),
    "random_best_params": {k: str(v) for k, v in random_search.best_params_.items()},
    "grid_best_cv_f2": float(grid_search.best_score_),
    "grid_best_std": float(best["std_test_score"]),
    "grid_best_min_fold": float(best["min_fold"]),
    "grid_best_params": {k: str(v) for k, v in grid_search.best_params_.items()},
    "n_tied_1std": int(len(tied)),
    "most_stable": {"params": {k: str(v) for k, v in most_stable["params"].items()},
                    "mean": float(most_stable["mean_test_score"]),
                    "std": float(most_stable["std_test_score"]),
                    "min_fold": float(most_stable["min_fold"])},
    "top10": top.astype(str).to_dict(orient="records"),
}
with open("results.json", "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=2, ensure_ascii=False)
print("Guardado results.json")
```
